Escribí una función reporte_semanal(ventas) que calcule e imprima:

El total vendido en la semana (con sum()).
El promedio diario (podés reusar tu función media, o hacerlo con sum()/len() — vos decidís).
El máximo y el mínimo del período (max(), min()).
Cuántos días superaron el promedio semanal — esto ya no lo resuelve una función built-in sola: necesitás calcular primero el promedio, y después un for que recorra ventas comparando cada valor contra ese promedio con un if, contando cuántas veces se cumple

```python
ventas = [1200, 950, 1500, 800, 2000, 1100, 1750]
```

```python
def total_vendido(lista):
  return sum(lista)
def media(lista):
  return sum(lista) / len(lista)


print(total_vendido(ventas))
media(ventas)
```

```python
def reporte_semanal(lista):
    print("El total vendido de la semana es:" , total_vendido(lista))
    print("El promedio dirio de la semana es:" , round(media(lista), 2 ))
    print("El dia que menos se vendio fue:", min(lista)," y el dia que mas se vendio fue :", max(lista))
    dias_arriba_promedio = 0
    media_reporte = media(lista)
    for i, dias in enumerate(lista):
      if lista[i] > media_reporte:
        dias_arriba_promedio = dias_arriba_promedio + 1
    print("La cantidad de dias arriba del promedio de ventas son:", dias_arriba_promedio)

reporte_semanal(ventas)
```
