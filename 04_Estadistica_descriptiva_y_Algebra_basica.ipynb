```python
def mediana(lista):
  if len(lista) % 2 == 1:
    return (sorted(lista)[len(lista)//2] )
  else :
    return ((sorted(lista)[len(lista)//2] ) + (sorted(lista)[len((lista)) //2 - 1] )) / 2
```

3) Escribí una función varianza(lista) que calcule la varianza (podés reutilizar tu función media de adentro). Probala con [5, 5, 5, 5] (debería dar 0, porque no hay dispersión) y con [0, 10, 0, 10].

```python
def varianza(lista):
  sumatoria = 0
  for i in lista:
    sumatoria += (i - media(lista)) **2
  return sumatoria / len(lista)
```

4) Escribí una función producto_punto(vector1, vector2) que reciba dos listas de igual longitud y devuelva su producto punto. Probala con precios = [4500, 500, 1500] y stocks = [25, 120, 44] — el resultado te tiene que dar la suma de los tres totales que ya calculaste antes (112500 + 60000 + 66000).

```python
precios = [4500, 500, 1500]
stocks = [25, 120, 44]

def producto_punto(lista1 , lista2):
  producto_total = 0
  for i in range(len(lista1)):
    acumulado = lista1[i]* lista2[i]
    producto_total += acumulado
  return producto_total
producto_punto(precios , stocks)
```
