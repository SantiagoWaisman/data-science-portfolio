```python
alumnos = [
    {"nombre": "Ana", "nota1": 8, "nota2": 7},
    {"nombre": "Luis", "nota1": 5, "nota2": 6},
    {"nombre": "Carla", "nota1": 9, "nota2": 10},
    {"nombre": "Pedro", "nota1": 4, "nota2": 5},
]

def procesar_alumnos(lista):
    promedio = 0 
    promedio_general= 0
    Alumnos_aprobados = []
    for alumno in lista:
        promedio = ((alumno["nota1"] + alumno["nota2"] )/2)
        alumno["promedio"] = promedio
        if promedio >= 6 :
            Alumnos_aprobados.append(alumno["nombre"])
        promedio_general +=  promedio / len(lista)
   



    return lista, Alumnos_aprobados, promedio_general



procesar_alumnos(alumnos)
```

```python
df = pd.DataFrame(alumnos)
print(df)
print(df.columns)
print(df.describe())
print(df.head(2))
```

```python
df_aprobados = df[df["promedio"]>= 6 ]
print(df_aprobados)
```
