```python
def es_mayor_de_edad(edad):
  if edad < 18:
    return "Menor de edad"
  elif edad == 18:
    return "18 recien cumplidos"
  else :
    return "Mayor de edad"
```

```python
es_mayor_de_edad(18)
```

```python
def es_mayor_de_edad_bool(edad):
    return edad >= 18
```

```python
es_mayor_de_edad_bool(19)
```

```python
def calcular_total(precio , stock):
  return (precio * stock)
```

```python
calcular_total(100,10)
```

```python
productos = ["Taza", "Goma", "Lapicera"]
precio = [4500, 500, 1500]
stock = [25, 120, 44]
```

```python
for i in range(len(productos)):
  print(calcular_total(precio[i] , stock[i]))
```
